```python
print("SOLVED EXAMPLE:")
print("Calculate the product of the first 10 natural numbers")

first_10_nat_nums = list(range(1, 11)) # Natural numbers start from 1
product_nat_nums = 1
for num in first_10_nat_nums:
    product_nat_nums *= num

print(f"First 10 natural numbers: {first_10_nat_nums}")
print(f"Product: {product_nat_nums}")
print("-" * 50)
```

```python
print("SOLVED EXAMPLE:")
print("Check if a number is a prime number")

num = 97
is_prime = True

if num <= 1:
    is_prime = False
elif num <= 3:
    is_prime = True
elif num % 2 == 0 or num % 3 == 0:
    is_prime = False
else:
    i = 5
    while i * i <= num:
        if num % i == 0 or num % (i + 2) == 0:
            is_prime = False
            break
        i += 6

if is_prime:
    print(f"{num} is a prime number")
else:
    print(f"{num} is not a prime number")
print("-" * 50)
```
